```python
from google.adk.agents import LlmAgent, SequentialAgent

print("Google ADK is ready!")
```

```python
import os
from getpass import getpass

# Prompt you to enter your Gemini API key
api_key = getpass("Enter your Gemini API key: ")

# Configure Google ADK
os.environ["GOOGLE_API_KEY"] = api_key
os.environ["GOOGLE_GENAI_USE_VERTEXAI"] = "FALSE"

print("Gemini API key configured successfully!")
```

```python
verifier_agent = LlmAgent(
    name="Verifier",
    model="gemini-3.8-flash",
    description="Checks simplified medical explanations for accuracy.",
    instruction="""
    You are a medical verification agent.

    Compare the simplified medical explanation
    against the original medical source.

    Check for:
    1. Incorrect medical information
    2. Unsupported claims
    3. Missing important facts
    4. Contradictions with the original source
    5. Changes in medical risk or uncertainty

    Explain any problems clearly so the
    Refiner Agent can correct them.
    """,
    output_key="verification"
)

print("Verifier Agent created!")
```
